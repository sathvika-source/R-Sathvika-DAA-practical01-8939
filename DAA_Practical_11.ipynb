```python
def floyd_warshall(graph):
    # Obtain list of all unique vertices
    vertices = list(graph.keys())
    n = len(vertices)

    # Map vertices to matrix indices
    v_to_idx = {v: i for i, v in enumerate(vertices)}

    # Initialize the distance matrix with infinity
    dist = [[float('inf')] * n for _ in range(n)]

    # Distance from a vertex to itself is 0
    for i in range(n):
        dist[i][i] = 0

    # Populate direct edge weights into the matrix
    for u in graph:
        for weight, v in graph[u]:
            dist[v_to_idx[u]][v_to_idx[v]] = weight

    # Core Floyd-Warshall dynamic programming step
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][j] > dist[i][k] + dist[k][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]

    return dist, vertices

# --- Example Usage ---
# Using the same graph structure (directed or converted representation for all-pairs)
example_graph = {
    'A': [(2, 'B'), (3, 'C')],
    'B': [(2, 'A'), (1, 'C'), (1, 'D'), (4, 'E')],
    'C': [(3, 'A'), (1, 'B'), (5, 'F')],
    'D': [(1, 'B'), (1, 'E')],
    'E': [(4, 'B'), (1, 'D'), (1, 'F')],
    'F': [(5, 'C'), (1, 'E')]
}

distances, vertex_order = floyd_warshall(example_graph)

print("Vertex order:", vertex_order)
print("Shortest distance matrix:")
for row in distances:
    print(row)
```
