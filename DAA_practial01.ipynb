## **Bubble Sort**

```python
import time

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):

        swapped = False
        for j in range(0, n - i - 1):

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break
    return arr

print("--- Bubble Sort ---")
example_arr_bubble = [64, 34, 25, 12, 22, 11, 90]
print(f"Original array: {example_arr_bubble}")

start_time = time.perf_counter()
sorted_arr_bubble = bubble_sort(example_arr_bubble.copy())
end_time = time.perf_counter()

print(f"Sorted array: {sorted_arr_bubble}")
print(f"Execution time: {(end_time - start_time) * 1000:.4f} ms\n")
```

## **Merge Sort**

```python
import time

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]

        merge_sort(L)
        merge_sort(R)

        i = j = k = 0


        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1


        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1
    return arr

print("--- Merge Sort ---")
example_arr_merge = [12, 11, 13, 5, 6, 7]
print(f"Original array: {example_arr_merge}")

start_time = time.perf_counter()
sorted_arr_merge = merge_sort(example_arr_merge.copy())
end_time = time.perf_counter()

print(f"Sorted array: {sorted_arr_merge}")
print(f"Execution time: {(end_time - start_time) * 1000:.4f} ms\n")
```

## **Insertion Sort**

```python
import time

def insertion_sort(arr):

    for i in range(1, len(arr)):
        key = arr[i]

        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr


print("--- Insertion Sort ---")
example_arr_insertion = [12, 11, 13, 5, 6]
print(f"Original array: {example_arr_insertion}")

start_time = time.perf_counter()
sorted_arr_insertion = insertion_sort(example_arr_insertion.copy())
end_time = time.perf_counter()

print(f"Sorted array: {sorted_arr_insertion}")
print(f"Execution time: {(end_time - start_time) * 1000:.4f} ms\n")
```

## **Quick Sort**

```python
import time

def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2]
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        return quick_sort(left) + middle + quick_sort(right)


print("--- Quick Sort ---")
example_arr_quick = [10, 7, 8, 9, 1, 5]
print(f"Original array: {example_arr_quick}")

start_time = time.perf_counter()
sorted_arr_quick = quick_sort(example_arr_quick.copy())
end_time = time.perf_counter()

print(f"Sorted array: {sorted_arr_quick}")
print(f"Execution time: {(end_time - start_time) * 1000:.4f} ms\n")
```

## **Selection Sort**

```python
import time

def selection_sort(arr):
    n = len(arr)

    for i in range(n):

        min_idx = i
        for j in range(i + 1, n):
            if arr[min_idx] > arr[j]:
                min_idx = j


        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


print("--- Selection Sort ---")
example_arr_selection = [64, 25, 12, 22, 11]
print(f"Original array: {example_arr_selection}")

start_time = time.perf_counter()
sorted_arr_selection = selection_sort(example_arr_selection.copy())
end_time = time.perf_counter()

print(f"Sorted array: {sorted_arr_selection}")
print(f"Execution time: {(end_time - start_time) * 1000:.4f} ms\n")
```
