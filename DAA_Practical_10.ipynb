```python
class DisjointSet:
    def __init__(self, vertices):
        self.parent = {v: v for v in vertices}
        self.rank = {v: 0 for v in vertices}

    def find(self, item):
        if self.parent[item] == item:
            return item
        # Path compression
        self.parent[item] = self.find(self.parent[item])
        return self.parent[item]

    def union(self, set1, set2):
        root1 = self.find(set1)
        root2 = self.find(set2)

        if root1 != root2:
            # Union by rank
            if self.rank[root1] < self.rank[root2]:
                self.parent[root1] = root2
            elif self.rank[root1] > self.rank[root2]:
                self.parent[root2] = root1
            else:
                self.parent[root2] = root1
                self.rank[root1] += 1
            return True
        return False

def kruskals_algorithm(graph):
    # graph is expected to be a dictionary:
    # {vertex: [(weight, neighbor), ...]}

    # 1. Collect all unique edges
    edges = []
    vertices = list(graph.keys())
    for u in graph:
        for weight, v in graph[u]:
            # Store edge normalized to avoid duplicates in undirected graph
            if u < v:
                edges.append((weight, u, v))

    # 2. Sort edges by ascending weight
    edges.sort()

    # 3. Initialize DSU and MST variables
    dsu = DisjointSet(vertices)
    mst = []
    total_weight = 0

    # 4. Process sorted edges
    for weight, u, v in edges:
        if dsu.union(u, v):
            mst.append((u, v, weight))
            total_weight += weight

    return mst, total_weight

# --- Example Usage ---
example_graph = {
    'A': [(2, 'B'), (3, 'C')],
    'B': [(2, 'A'), (1, 'C'), (1, 'D'), (4, 'E')],
    'C': [(3, 'A'), (1, 'B'), (5, 'F')],
    'D': [(1, 'B'), (1, 'E')],
    'E': [(4, 'B'), (1, 'D'), (1, 'F')],
    'F': [(5, 'C'), (1, 'E')]
}

mst_kruskal, total_cost_kruskal = kruskals_algorithm(example_graph)
print("Minimum Spanning Tree (Edges):", mst_kruskal)
print("Total MST Weight:", total_cost_kruskal)
```
